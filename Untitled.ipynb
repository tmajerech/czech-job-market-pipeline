```python
from pathlib import Path
import gzip
import json
import pandas as pd

FULL_JSON = Path(
    "data/raw/vacancies/full/"
    "volna-mista-retrieved-2026-10-03.json.gz"
)

with gzip.open(FULL_JSON, mode="rt", encoding="utf-8") as file:
    raw_data = json.load(file)

df = pd.json_normalize(raw_data["polozky"], sep=".")

print(df.shape)
df.head()
```
